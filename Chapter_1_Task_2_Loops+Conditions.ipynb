## 3. if else amd if elif else

```python
user_input_if = int(input("Enter a number:"))
if (user_input_if <= 5):
    print("Value is less than or equal to {}".format(user_input_if))
else:
    print("User input range is more than 5")
```

```python
user_input_if = int(input("Enter a number:"))
if (user_input_if <= 5):
    print("Value is less than or equal to {}".format(user_input_if))
elif (user_input_if > 5 and user_input_if <= 15):
    print("User input range is between 6-15")
else:
    print("User input greater than 15")
```

## 3. User defined function with for & if (break)

```python
def print_user_input_range(user_input):
    for i in range(user_input):
        print(i)

print_user_input_range(3)
```

```python
def print_only_even_numbers(user_input):
    for i in range(user_input):
        if i%2==0:
            print(i)
        else:
            continue
   
    print("------------------------") 

    for i in range(user_input):
        if i==5:
            break
        else:
            print(i)
print_only_even_numbers(25)
```

## 3. for loop with in operator & not in

```python
buying_list = ["Bread", "Milk", "chips", "Ketchup", "Butter", "chocolate"]
skip_junk_item = ["chips", "chocolate"]

for item in buying_list:
    print("Item: ", item)

for item in buying_list:
    if item not in skip_junk_item:
        print("Buy_item: ", item)
    else:
        print("Junk_item: ", item)
```
