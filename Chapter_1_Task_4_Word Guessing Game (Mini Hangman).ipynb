## Word Guessing Game (Mini Hangman)

```python
import random
print("Welcome to Word Guess Game !!!")

def get_user_input():
    while True:
        try:
            user_input = input("Enter a letter to guess the fruit name: ")

            if len(user_input) == 1 and user_input.isalpha():
                return user_input
            else:
                print("Input is Invalid, please enter only single alphabet (Ex: a or b or c).")
        except ValueError:
            print("Input is Invalid, please enter only single alphabet (Ex: a or b or c).")

###################################################################################################

def place_char_placeholder(user_char_list, system_word):
    #print(user_char_list, system_word)
    found = True
    for sys_char in system_word:
       if sys_char.lower() in [c.lower() for c in user_char_list]: # this line bit tricked 
           print(sys_char, end=" ")
       else:
           print("_", end=" ")
           found = False
    return found

###################################################################################################
    
words = ["Orange", "Mango", "Apple"]
system_word = words[random.randint(0,2)]
user_char_list = []

wrong_guesses = 0
max_try = 6
print()

while True:
    #print(system_word)
       
    print("Wrong attempts : {} ({})".format(wrong_guesses, max_try))
    print()
    user_char = get_user_input()
    
    
    if user_char.lower() in [c.lower() for c in user_char_list]:
        print(f"You have already gussed this character '{user_char}'")
        continue
    
    if system_word.lower().find(user_char.lower()) == -1:
        wrong_guesses += 1
        print("Wrong guess !!!")

        if wrong_guesses >= max_try:
            print("You lost your chances and the word is {}".format(system_word))
            break
        continue   
    
    user_char_list.append(user_char)
   
    found = place_char_placeholder(user_char_list, system_word.strip())
    print()

    if found == True:
        print("You Won !!! Have a Good Day !!!!!")
        break

    
#sys_word_length, system_word, user_char_list

```
