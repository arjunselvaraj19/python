## Merge fucntion

```python
# creating inputs

student_primary_details = {"Student_ID": [1001,1002,1003,1004,1005],
                            "Name":["Arjun","Mithila","Preethi", "Rithvik", "Billa"],
                            "Age": [40,2,35,12, 45],
                            "Location": ["Namakkal","Germany","Chennai","Doha", "Delhi"]}

student_education_details = {"Student_ID": [1001,1002,1003,1004,1006],
                             "Study": ["B.E","PHd","Msc","MBA", "Diploma"],
                             "Duration of Course": [4,2,3,3, 2]}

student_language_details = {"Student_ID": [1001,1002,1003,1008],
                            "Name":["Arjun","Mithila","Preethi", "Rithvik"],
                            "Language": ["English","German","Telugu","Arabic"],
                            "Mother Tongue": ["Tamil", "English", "Tamil", "German"]}

import pandas as pd
stud_primary_df = pd.DataFrame(data=student_primary_details)
stud_education_df = pd.DataFrame(data=student_education_details)
stud_language_df =  pd.DataFrame(data=student_language_details)
```

```python
#type(stud_language_df)
#Syntax

# output = pd.merge(left, right, how, on)

# main parameters
'''
left - Left side DataFrame
right - Right  "
how - what type of join keyword
on - On which column to join (more than 1 col also applied)
'''
```

## Left Join
- how : left
- left side all records appeared if any miss on right it will be mentioned as NaN

```python
pd.merge(stud_primary_df, stud_language_df, how="left", on=["Student_ID", "Name"])
```

## Right Join
- how : right
- opposite to left (all rows from right side and matching from left. If no match then Nan)

```python
pd.merge(stud_primary_df, stud_education_df, how="right", on=["Student_ID"])
```

## Inner Join
- Only matched bet both side

```python
pd.merge(stud_primary_df, stud_language_df, how="inner", on=["Student_ID", "Name"])
```

## Outer Join
- all rows from both side and if any non match will Nan

```python
pd.merge(stud_primary_df, stud_language_df, how="outer", on=["Student_ID", "Name"])
```
