```python
# importing sales datas
import pandas as pd
sales_data_2017 = pd.read_csv(filepath_or_buffer="Sales Transactions-2017.csv")
sales_data_2018 = pd.read_csv(filepath_or_buffer="Sales Transactions-2018.csv")
sales_data_2019 = pd.read_csv(filepath_or_buffer="Sales Transactions-2019.csv")

# Combining using concat
sales_data_full = pd.concat(objs= [sales_data_2017, sales_data_2018, sales_data_2019])
```

```python
### Party can be anybody but product & Qty is same. 
#   So here we will find out the mean with the above filter and apply on the NaN
sales_data_filtered = sales_data_full[(sales_data_full["Product"] == "TIRML-BL-KG-17*23") & (sales_data_full["Qty"] == "2")]

## Needs to be changed the datatype and need to remove empty for aggregation
sales_data_filtered_withoutNA = sales_data_filtered[~sales_data_filtered["Rate"].isna()]
sales_data_filtered_withoutNA
```

```python
sales_data_filtered_withoutNA.dtypes
```

```python
## assign datatype for Rate field
sales_data_filtered_withoutNA["Rate"] = sales_data_filtered_withoutNA["Rate"].astype(float).copy()
sales_data_filtered_withoutNA.dtypes
```

```python
#finding mean value 
mean_value = sales_data_filtered_withoutNA.groupby(by=["Product"])["Rate"].aggregate(func=["mean"])
mean_value
#sales_data_filtered.loc[sales_data_filtered[]]
```
