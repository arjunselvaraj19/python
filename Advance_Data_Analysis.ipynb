### 1. Groupby Function

```python
import pandas as pd
```

```python
insurance_data = pd.read_csv("Insurance.csv")
insurance_data
```

#### 1.1 Perform Initial Analysis

```python
insurance_data.isna().sum()
```

```python
insurance_data.dtypes
```

```python
insurance_data.describe()
```

#### 1.2 Pull up the Average Insurance chanrges based on region

```python
insurance_data["region"].unique()
```

```python
insurance_data.groupby(by="region")["charges"].mean().round(2).sort_values()
```

#### 1.3 Pull up the Average Insurance chanrges based on region and sex

Discrete and Continous data

* Discrete - Just Countable / Cant' be segmented (rest all comes under this)
* Continous - Measurable / Have units / can be segmented (Age, charges)

```python
insurance_data.groupby(by=["region", "sex"])["charges"].mean().round(2).sort_values()
```

### 2. Pivot Table

```python
# pivot table more presentable comparing 2 categorical features vs 
pd.pivot_table(data=insurance_data, values="charges", index="region", columns=["sex","children"]).round(2)
```

### 3. Aggregate

```python
insurance_data.head()
```

### 3.1 Find out the avergae insurance charges based on Region

Common aggregate functions are min, max, sum and mean

```python
insurance_data.groupby(by="region")["charges"].mean().sort_values().round(2)
```

### 3.2 Find out the aggregae of min, max, sum & mean

```python
#insurance_data.groupby(by=["region"])["charges"].aggregate(func=["mean", "min", "max", "sum"]).round(2)
insurance_data.groupby(by=["region","sex"])["charges"].aggregate(func=["mean", "min", "max", "sum"]).round(2)
```

## 6. Apply

### Create a new column in the same Dataframe with the name "Elibility criteria" where if age is > 20 

```python
def get_age_eligibility(age):
    if age > 20:
        return "Eligible"
    else:
        return "Not eligible"

insurance_data["Eligible Criteria"] = insurance_data["age"].apply(get_age_eligibility)
insurance_data
```

## 7. Lambda function
- Anonymous Function / Nameless function

```python
insurance_data["Eligible Criteria_1"] = insurance_data["age"].apply(lambda x:True if x > 20 else False)
insurance_data
```
