## Lambda Function with Map, Filter, Reduce

### 1. map
- it returns same map type object so we need convert as list

```python
numbers = [0,5,10,15,20,25,30,35,40,45,50,55,60]

resultt = list(map(lambda x: x*2, numbers))
print(resultt)
```

```python
names = ["abc","xyz","ijk"]
resultt = list(map(lambda x: x.upper(), names))
print(resultt)
```

```python
names = ["Arjun","Ramesh","RoopKumer"]
nick_name = ["AJ","RAM", "RK"] # if any mismatch the length it skips

resultt = list(map(lambda n, nn : f"{n} alias {nn}", names, nick_name))
print(resultt)
```

### 2. filter

```python
fruits = ["Mango Fruit","Apple Fruit","Lion Animal","Peach Fruit","Tiger Animal","Pomegranate Fruit"]

resultt = list(filter(lambda x: "Fruit" in x, fruits))
print(resultt)
```

```python
numbers = [-5, -3, 0,2, 4, -1,7]

resultt = list(filter(lambda x: x > 0, numbers))
print(resultt)
```

```python
resultt = list(filter(lambda x : x.endswith("mal"), fruits))
print(resultt)
```

### reduce
- return one values

```python
# reduce is not a buitl-in, need to pull functools
from functools import reduce

numbers = [5,10,15,20,25]

resultt = reduce(lambda a,b: a+b, numbers)
print(resultt)
```

```python
names = ["Arjun","Ramesh","RoopKumer"]

resultt = reduce(lambda a, b: a + b, names)
print(resultt)
```
