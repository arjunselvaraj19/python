## 3. Methods
1. Instance Method
    - 1st parameter is self
    - can access and modify instance vars
3. Class Method
    - Above method uses **@classmethod**
    - 1st parameter is cls
    - can access and modify class vars
    - called by class, not an object (need to verify
5. Static Method
    - Above method uses **@staticmethod**
    - no self or cls parameter
    - cannot access or modify any variables

```python
class Employee:
    company_name = "TechCorp"   # class variable

    def __init__(self, name, salary):
        self.name = name         # instance variable
        self.salary = salary

    # 1. Instance Method - works with self (object data)
    def display(self):
        print(f"Name: {self.name}, Salary: {self.salary}, Company: {self.company_name}")

    # 2. Class Method - works with cls (class data), here you cannot access self vars
    @classmethod
    def change_company(cls, new_name):
        cls.company_name = new_name
        print(f"Company changed to: {cls.company_name}")

    # 3. Static Method - no self, no cls
    @staticmethod
    def validate_salary(salary):
        company_name = "adsasd" #this is not affect the objects values
        if salary > 10000:
            print(f"{salary} is a valid salary")
        else:
            print("Salary must be greater than 10000")


# Creating objects
emp1 = Employee("Arjun", 70000)
emp2 = Employee("Rambabu", 45000)

# Instance method - called on object
emp1.display() 
emp2.display()  
print()
# Class method - called on class (affects all objects)
Employee.change_company("NewTech")
#print(Employee.company_name)  
print()
emp1.display() 
emp2.display() 
print()

# Static method - called on class, no instance needed directly call using class name
Employee.validate_salary(50000)  
Employee.validate_salary(5000) 
print()
emp1.display() 
emp2.display() 
```

## 4. Encapsulation
- means hiding internal data and controlling how its accessed or moddified from outside the class

  **3 Levels of Access**
  1. Public - access anywhere (self.name)
  2. Protected - Class + Subclasses (self._salary)
  3. Private - Only inside the class (self.__password)

```python
class BankAccount:
    def __init__(self, owner, balance, password):
        self.owner = owner          # Public - anyone can access
        self._balance = balance     # Protected - internal use / subclasses
        self.__password = password  # Private - strictly inside this class only

    # Public method to deposit
    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            print(f"Deposited {amount}. New balance: {self._balance}")

    # Public method to withdraw
    def withdraw(self, amount, password):
        if password != self.__password:
            print("Wrong password! Access denied.")
        elif amount > self._balance:
            print("Insufficient funds!")
        else:
            if amount > 0:
                self._balance -= amount
                print(f"Withdrawn {amount}. New balance: {self._balance}")
            else:
                print("Please provide the valid amout!!")

    # Public method to check balance (controlled access)
    def get_balance(self, password):
        if password == self.__password:
            return self._balance
        else:
            print("Wrong password!")
            return None

#class JointBankAccount(BankAccount):
    

# Create account
acc_Arjun = BankAccount("Arjun", 5000, "secret123")
acc_Mithila = BankAccount("Mithila", 10000, "secret456")

# Public - works fine
print(acc_Arjun.owner)

# Protected - accessible and able to módify also
print(acc_Arjun._balance)        
acc_Arjun._balance = 15000
print(acc_Arjun._balance)

# Private - can'T able to access the __var(protected)
## print(acc_Arjun.__password)

# Use methods instead (correct way)
acc_Arjun.deposit(1000)          # Deposited 1000. New balance: 6000
#acc_Arjun.withdraw(500, "secret123")    # Withdrawn 500. New balance: 5500
acc_Arjun.withdraw(500, "wrongpass")    # Wrong password! Access denied.
print("Balance:",acc_Arjun.get_balance("secret123")) #16000
```

## 5. Inheritence
- child class reuse the code of parent class

  1. Single Inheritence (1 parent --> 1 child)
  2. Multilevel Inheritence (Grandparent --> PArent --> Child)
  3. Hierarchical Inheritence (1 Parent --> many Childs)
  4. Multiple Inheritence (1 child --> man PArents)

```python
class Coder:
    def code(self):
        print("Writing code")

class Tester:
    def test(self):
        print("Testing code")

class FullStackQA(Coder, Tester):    # inherits from both
    def deliver(self):
        print("Delivering tested code")


person = FullStackQA()
person.code()
person.test()
person.deliver()
```

### Using Super keyword
- It is used to call the parent constructor / method from sub class
- when you use Super for constr or method then the method name should be available in parent

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def display(self):
        print(f"Name: {self.name}")


class Developer(Employee):
    def __init__(self, name, language):
        super().__init__(name)          # 1. reuse parent constructor
        self.language = language

    def display(self):
        super().display()               # 2. run parent method first
        print(f"Language: {self.language}") 


Developer("Arjun", "Python").display()
```

```python
dev = Developer("Arjun", "Python")

print(isinstance(dev, Developer))
print(isinstance(dev, Employee))
print(issubclass(Developer, Employee))
print(issubclass(Employee , Developer))
```

## 6. Polymorphism
- means many forms

### Method Overriding

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def work(self):
        print(f"{self.name} is working")


class Developer(Employee):
    def work(self):            # overrides parent's work()
        print(f"{self.name} is writing code")


class Tester(Employee):
    def work(self):                          # overrides parent's work()
        super().work() # this hits parent first.
        print(f"{self.name} is testing the application")


class Manager(Employee):
    pass         # no override → uses parent's work()


dev = Developer("Arjun")
qa = Tester("Rambabu")
mgr = Manager("Priya")

dev.work() 
qa.work() 
mgr.work() 
```

## 7. Abstraction
- means showing only what user needs and hiding
- defining a set of rules that every child class must follow.

**abc module**
- Python provides the **abc** module (Abstract Base Classes)
  * **ABC**: inherit from it to make a class abstract
  * **@abstractmethod**: makes a method that every child **must** implement

```python
from abc import ABC, abstractmethod


class Payment(ABC):   # class inherits base absract class (ABC) so Payment becomes abstract class
    def __init__(self, amount):
        self.amount = amount

    @abstractmethod #every child class must have pay method
    def pay(self):
        pass

    def receipt(self):  # Normal method: shared by all children
        print(f"Receipt: {self.amount} EUR paid successfully")


class CreditCard(Payment):
    def pay(self):
        #self.amount = self.amount
        print(f"Paid {self.amount} EUR using Credit Card")


class PayPal(Payment):
    def pay(self):
        print(f"Paid {self.amount} EUR using PayPal")


class Investment(Payment):
    def pay(self): # If I rename the pay method as payment --> 'Can't instantiate abstract class Investment with abstract method pay'
        self.amount = self.amount * 2
        print(f"Paid {self.amount} EUR")

class Loan(Payment):
    def pay(self):
        pass


payments = [CreditCard(100), PayPal(250), Investment(75), Loan(800)]

for p in payments:
    p.pay()
    p.receipt()
    print()

# defining abc mrthods in child class is complusory but yu can have it as empty (mentioning pass)
```

```python
from abc import ABC, abstractmethod


# Level 1: Abstract parent (sets the rules)
class Employee(ABC):
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @abstractmethod
    def work(self):
        pass

    @abstractmethod
    def calculate_bonus(self):
        pass

    # Concrete method: shared by everyone, and it uses the abstract method
    def display(self):
        print(f"{self.name} | Salary: {self.salary} | Bonus: {self.calculate_bonus()}")


# Level 2: Defines ONLY ONE abstract method → still abstract
class TechEmployee(Employee):
    def calculate_bonus(self):
        return self.salary * 0.10 


# Level 2: Defines BOTH → concrete
class Manager(Employee):
    def work(self):
        print(f"{self.name} is managing the team")

    def calculate_bonus(self):
        return self.salary * 0.20


# Level 3: Defines the remaining work() → concrete
class Developer(TechEmployee):
    def __init__(self, name, salary, language):
        super().__init__(name, salary)     # goes up to Employee.__init__
        self.language = language

    def work(self):
        print(f"{self.name} is coding in {self.language}")


# Level 3: Defines work() AND overrides the bonus using super()
class Tester(TechEmployee):
    def work(self):
        print(f"{self.name} is testing the application")

    def calculate_bonus(self):
        return super().calculate_bonus() + 1000   # 10% from parent + 1000 extra


staff = [
    #TechEmployee("Rithvik", 70000), this you cann't create bcoz there is no method defined for work
    Developer("Arjun", 70000, "Python"),
    Tester("Rambabu", 45000),
    Manager("Priya", 90000),
]

for emp in staff:
    emp.work()
    emp.display()
```
