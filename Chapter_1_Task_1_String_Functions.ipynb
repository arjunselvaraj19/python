Examples of all string functions

```python
customer_name = "ARJUN    Selvaraj"
customer_nic_name = "  aRJUn    "
customer_place = "mini Japan japan"
customer_contact_number = "56646-564646486867878"
customer_filename = "Cover_Letter1919.pdf"   

print("****** capitalize() ******") # makes first char capital and rest all lower case
print(customer_name.capitalize())
print(customer_place.capitalize())
print()

print("****** casefold() ******") # makes everything lower case
print(customer_name.casefold())
print(customer_place.casefold())
print()

print("****** center() ******") # makes the string center by adding the padding on both side
print("'",customer_name.center(20),"'")
print("'",customer_place.center(20),"'")
print("'",customer_place.center(5),"'")
print()

print("****** count() ******") # provide the avialable occurence
print(customer_name.count("a"))
print(customer_name.count("A"))
print(customer_place.count("i"))
print(customer_place.count("ap"))
print()

print("****** encode() ******") # encodes the string
print(customer_name.encode())
print(customer_place.encode())
print()

print("****** endswith() & startswith() ******") # returns bool when the string matches with input (both direction depends on keyword)
print(customer_name.endswith("raj"))
print(customer_name.endswith("Arjun"))
print(customer_place.endswith(""))
print(customer_filename.endswith(".pdf"))
print(customer_filename.endswith("Letter"))
print(customer_name.startswith("raj"))
print(customer_name.startswith("Arjun"))
print(customer_filename.startswith(".pdf"))
print(customer_filename.startswith("Letter"))
print()

print("****** expandtabs() ******") # Used to replace the Tab or apply the tabs
csv_header = "ID\tName\tAge\tGender\t\tLocation\t\tDOB\n1001\tMathesh\t30\tMale\t\tMumbai\t\t30Sep2010"
print(csv_header.expandtabs()) # default tab size is 8
print(csv_header.expandtabs(20)) # it is custom
print()

print("****** find() & rfind() ******") # used to find char or straing 
print(customer_name.find("Sel")) # returns lowest index (starting char) when matched
print(customer_place.find("jjj")) # if not then returns -1
print(customer_place.find("j")) # returns lowest or first finding index
print(customer_place.rfind("j")) # returns highest or last finding index
print(customer_place.rfind("an")) # returns highest or last finding index
print()

print("****** format() ******") # Used to insert the values into placeholder
print("India is a {} country in {}".format("Great", "Asia")) # applying values in placeholder
print("This is my {} trip in {}".format("first", "Asia")) # applying values in placeholder
print()

print("****** format_map() ******") # It is also silmilar like frmat but it covers like render concept
person = {"name":"Arjun","age": 32,"job":"farmer"}
print("NAme: {name}, Age: {age}, activity: {job}".format_map(person)) # it fetches vlue from person and applies into placeholder
print("Employee_name : {name}, bms_value: {age}, hobby: {job}".format_map(person)) # it fetches vlue from person and applies into placeholder
print()

print("****** index() & rindex() ******") # used to find the index of substring on both direction
print(customer_name, customer_name.index("S")) # returns the index number
print(customer_place, customer_place.index("jap")) # returns the last matched lowest index.
print(customer_place, customer_place.rindex("jap")) # returns the last matched lowest index.
print(customer_place, customer_place.rindex("an")) # returns the last matched lowest index.
print()

print("****** isalnum() ******") # used to find the string is only alphanumeric
var_isalnum = "adsdf21pdf"
print(var_isalnum, var_isalnum.isalnum()) # if pure alphanum then returns bool true
print(customer_place, customer_place.isalnum()) # if any space or spl char available then False
print(customer_filename, customer_filename.isalnum()) # spl or dot or comma 
print()

print("****** isalpha() ******") # used to find the string is only alpha or string retun type is bool
var_isalphs = "abcdABCD"
print(customer_name, customer_name.isalpha())
print(customer_filename, customer_filename.isalpha())
print(var_isalphs, var_isalphs.isalpha())
print()

print("****** isascii() ******") # used to find the string is ASCII (return true or false) 
var_isascii = "abcdßß"
print(customer_name, customer_name.isascii()) # if string contains alha numberic or spl chars thens it returns True
print(customer_place, customer_place.isascii())
print(var_isascii, var_isascii.isascii()) # if string contains and non ascii content then false.
print()

print("****** isdecimal() ******") # used to find the string as decimal 
print("283", "283".isdecimal()) # if pure decimal the true
print("28.3", "28.3".isdecimal()) # if float then false
print("-283", "-283".isdecimal()) # if negative the false
print("283G", "283G".isdecimal()) # if alphanum then false
print()

print("****** isdigit() ******") # used to find the string as decimal or digit - similar to isdecimal but it handles special digits 
print("283", "283".isdigit()) # if pure decimal the true
print("28.3", "28.3".isdigit()) # if float then false
print("-283", "-283".isdigit()) # if negative the false
print("283G", "283G".isdigit()) # if alphanum then false
print("①②③", "①②③".isdigit()) # if spl digits then true  ********
print("1/2", "1/2".isdigit()) # if alphanum then false
print()

print("****** isidentifier() ******") # used to find the string as identifier (whether it could be single or combinational)
print("def", "def".isidentifier()) # def is indentifier so True
print("while-for","while-for".isidentifier()) # while and for is also the indentifier but it has separated with hypen so false. it considers as non.identifier.
print("while_for_elif", "while_for_elif".isidentifier()) # combination of identifier but separate with underscore it allows so true
print("dowhileifelseelif", "dowhileifelseelif".isidentifier()) # combination also allows do True
print()

print("****** islower() ******") # used to find all the chars are lower case
var_islower = "ab.c1d*ef gh"
print(var_islower, var_islower.islower()) # return True when string has all lower alpha or alphanum or with spl char also 
print(customer_name, customer_name.islower()) # returns false when string has even single CAPITAL char
print(customer_place, customer_place.islower())
print(customer_contact_number, customer_contact_number.islower())
print(customer_filename, customer_filename.islower())
print()

print("****** isnumeric() ******") # used to find the complete string as numeric 
print("23432423", "23432423".isnumeric()) # if it is pure numeric then true
print("23432423.75", "23432423.75".isnumeric()) # float - false
print("-23432423", "-234324235".isnumeric()) #  negative - false
print("-23432423.75", "-23432423.75".isnumeric()) # negative, float - false
print("0", "0".isnumeric()) # 0 - true
print()

print("****** isprintable() ******") # used to find the complete string as printable (allows empty string<
print(customer_name, customer_name.isprintable()) 
print(customer_filename, customer_filename.isprintable())
print("abcdef\t123", "abcdef\t123".isprintable())
print("abcdef\n12345", "abcdef\n12345".isprintable())
print()

print("****** isspace() ******") # used to find the string has only empty space or \t or \n,...
print(customer_name, customer_name.isspace()) 
print(customer_place, customer_place.isspace())
print(customer_filename, customer_filename.isspace())
print("", "".isspace())
print(" ", " ".isspace())
print("\t", "\t".isspace())
print()

print("****** istitle() ******") # used to find the string has Title case pattern (Each word or char should starts with upper case char)
var_istitle = "My Name Is Little Panther"
print(var_istitle, var_istitle.istitle())
print(customer_name, customer_name.istitle()) 
print(customer_place, customer_place.istitle())
print(customer_contact_number, customer_contact_number.istitle())
print(customer_filename, customer_filename.istitle())
print()

print("****** isupper() ******") # used to find all the string content is upper case.
var_istitle = "MY NAME"
print(var_istitle, var_istitle.isupper())
print(customer_name, customer_name.isupper()) 
print(customer_place, customer_place.isupper())
print(customer_contact_number, customer_contact_number.isupper())
print(customer_filename, customer_filename.isupper())
print()

print("****** join() ******") # used to join the strings and can be formatted like includes spl chars
va_join = ["Hi", "How","are","you","doing"]
print(va_join, " ".join(va_join))
print(va_join, ", ".join(va_join))
print(va_join, "@ ".join(va_join))
print()

print("****** ljust() & rjust() ******") # used to justify the string - indent
print(customer_place.ljust(20), "'")
print(customer_place.rjust(20), "'")
print()

print("****** lower() & upper() ******") # used to apply the upper or lower case completely
print(customer_place, customer_place.lower())
print(customer_place, customer_place.upper())
print()

print("****** lstrip() & rstrip() & strip() ******") # used to trim the empty or white spaces left or rgith or both (depends on keyword)
print(customer_nic_name, customer_nic_name.lstrip())
print(customer_nic_name, customer_nic_name.rstrip())
print()

print("****** maketrans() & translate() ******") 
# used to allocate the equivalent chars like For 'a' is 'x' and then by using translate we can tranlated the letters
var_translation = str.maketrans("abc","xyz")
print(str, var_translation)
print("abc".translate(var_translation))
print(customer_nic_name, customer_nic_name.translate(var_translation))
print()

print("****** partition() & rpartition() ******") # used to find the first match and partition into 3 parts as tuple
var_partition = "Cover_Letter1919_pdf"
print(customer_filename, customer_filename.partition("."))
print(var_partition, var_partition.partition("_"))
print(var_partition, var_partition.rpartition("_"))
print()

print("****** removeprefix() & removesuffix() ******") # used to trim the specified string on both direction
print(customer_filename, customer_filename.removeprefix("Cover"))
print(customer_place, customer_place.removeprefix("Mi"))
print(customer_place, customer_place.removeprefix("mi"))
print(customer_filename, customer_filename.removesuffix("pdf"))
print(customer_place, customer_place.removesuffix("pan"))
print(customer_place, customer_place.removesuffix("paN"))
print()

print("****** replace() ******") # used to replace the specified content in string
var_replace = "Cover_Letter1919_pdf"
print(var_replace, var_replace.replace("_", "@").replace("er","AR"))
print()

print("****** split() & rsplit() ******") # used to split a string and returns as list (right or left depends on keyword)
var_fruits = "apple,banana,orange,grape"
print(var_replace, var_replace.split("_")) #left
print(var_fruits, var_fruits.split(",", 2)) #left
print(var_replace, var_replace.rsplit("_")) #right
print(var_fruits, var_fruits.rsplit(",", 2)) #right
print()

print("****** splitlines() ******") # used to split the line break formats
var_splitlines = "How\nare\nyou\ndoing"
print(var_splitlines, var_splitlines.splitlines())
print()

print("****** swapcase() ******") # used to flip the case (lower to upper and upper to lower)
var_swapcase = "How are you doing"
print(var_swapcase, var_swapcase.swapcase())
print()

print("****** title() ******") # used to titlize like makes first char as upper for each word
var_title = "How are you doing"
print(var_title, var_title.title())
print()

print("****** zfill() ******") # used to add the padding as '0' from left
var_title = "How are you doing"
print(var_title, var_title.zfill(25))
print()

### string function end
```
