### Default Parameter
> Default parameter should be last in list

```python
def addition(arg1, arg2, arg3=0):
    return arg1 + arg2 + arg3
addition(32,65)
```

###  Variable length argument
Cannot have more than one Var length arg

```python
def get_details(name, place, *expexted_price):
    print("Client_name: ", name)
    print("Place: ", place)
    print("Expected Price: ", expexted_price)
```

```python
get_details("Ramkumer","Chennai",50.8,63.9)
# *arg - here you can pass multiple values
```

### Keyword Argument
Cannot have more than one Var length arg

```python
def get_user_details(name, place, **expexted_price):
    print("Client_name: ", name)
    print("Place: ", place)
    print("Expected Price: ", expexted_price)
```

```python
get_user_details("Ramkumer","Chennai", ask_1 = 50.8, ask_2 = 63.9, ask_3 = 28)
```

## 5.4 Challenge-2 Find the Factors of a number
If user enters 20, o/p should be "The factors of 20 is 1, 2,4,5,10,10"

```python
user_input_num = int(input("Enter Number"))

for i in range(1, user_input_num+1):
    if user_input_num%i==0:
        print(i)
```

```python
user_input_num1 = int(input("Enter Number"))

result = []
for i in range(1, user_input_num1 + 1):
    if user_input_num1%i==0:
        result.append(i)

print("The factors of {} is {}".format(user_input_num1, result))
```
