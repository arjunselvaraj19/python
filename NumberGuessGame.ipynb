# Number Guessing Game

##  Level - 1 : Print if the user guessed no is right or not.

```python
secretNumber = 56
user_input = int(input("Enter your guess no:"))
if secretNumber == user_input:
    print("Congrats - your guess is Right!")
else:
    print("Sorry - Your guess is wrong")
```

##  Level - 2 : Print if the user guessed no is right or not but with max os 'N' trails

#### 2.1 for loop - end iteration is defined

```python
secretNumber = 56
for i in range(0,5):
    user_input = int(input("Enter your guess no:"))
    if secretNumber == user_input:
        print("Congrats - your guess is Right!")
        break
    else:
        print("Sorry - Your guess is wrong")
```

##  Level - 3 : User should provide only between 1 to 100. Output should like your is less than secret number or higher.

```python
print("Welcome to Number Guess Game")
print("Guess the number bet 1 - 100")

secret_number = 77

for i in range(1,8):
    user_input = int(input("Guess your number: "))

    if user_input < 1 or user_input > 100:
        print("User input is Invalid, please provide between 1 -100")
    else:
        if user_input == secret_number:
            print("Contrats - your guess is Correct !!!")
            print(f"Contrats - you have guessed the {secret_number} in {i} attempts!!!")
            break
        elif user_input < secret_number:
            print("Less than secret number")
        else:
            print("Greater than secret number")
```

##  Level - 5 : Avoid string input from user, should enter only integer

```python
import random

print("Welcome to Number Guess Game")
print("Guess the number bet 1 - 100")

secret_number = random.randint(1,100)

for i in range(1,8):
    try:
        user_input = int(input("Guess your number: "))
        if user_input < 1 or user_input > 100:
            print("User input is Invalid, please provide between 1 -100")
        elif user_input > secret_number:
            print("Greater than secret number")
        elif user_input < secret_number:
            print("Less than secret number")
        else:
            print("Your guess is Right!!")
            break
    except ValueError:
        print("Invalid Input, Not recognized. please provide between 1 -100")


if user_input == secret_number:
    print(f"Congratulations, You have guessed the number {secret_number} in {1} attempts.")
else: 
    print("Sorry, your trails are over!! Try Again")
```

##  Level - 5 : User defined function - run as a function.

```python
def number_guessing_game():
    import random
    
    print("Welcome to Number Guess Game")
    print("Guess the number bet 1 - 100")
    
    secret_number = random.randint(1,100)
    
    for i in range(1,8):
        try:
            user_input = int(input("Guess your number: "))
            if user_input < 1 or user_input > 100:
                print("User input is Invalid, please provide between 1 -100")
            elif user_input > secret_number:
                print("Greater than secret number")
            elif user_input < secret_number:
                print("Less than secret number")
            else:
                print("Your guess is Right!!")
                break
        except ValueError:
            print("Invalid Input, Not recognized. please provide between 1 -100")
    
    
    if user_input == secret_number:
        print(f"Congratulations, You have guessed the number {secret_number} in {1} attempts.")
    else: 
        print("Sorry, your trails are over!! Try Again")
```
