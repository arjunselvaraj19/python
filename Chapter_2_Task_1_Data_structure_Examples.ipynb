### 1. List
- 1D
- Heterogenous
- Accepts Duplicates values
- Sliced
- Mutable
- Empty var = []
- Value assiging = ["abc", "xyz"]

```python
var_list = [] # Creating empty varaible
#type(var_list)
var_list =["Chemistry","Maths","Physics","History","Drawing", "Maths", "G","h", 7, 34, 98, 67.7, 45.23, 0, -9, -78] # Assigning values to var
print("List[] : ", var_list)
print()
###########################################################################

print("******   append()   *****   ")
var_list.append("Tree") # Append a single value
var_list.append(["Leaf","Vitamins"]) # Assigning an another List - intersting
print("append :", var_list)
print()

print("******   clear()   *****   ")
var_list.clear() # Clear everything inside list
print("clear :", var_list)
print()
var_list = ["Chemistry","Maths","Physics","History","Drawing","G","h", "h", 7, 34, 98, 67.7, 45.23, 0, -9, 7, -78, -9, "h"] # Assigning values to var

print("******   copy()   *****   ")
var_list2 = var_list.copy() # Copy the values to new variable but not ref.
print("copy :", var_list2)
print()

print("******   count()   *****   ")
# Count provides the number of occurences in List
print("count (-9):", var_list.count(-9))
print("count (7):", var_list.count(7))
print("count (h):", var_list.count("h"))
print("count (mh):", var_list.count("mh"))
print("count (Maths):", var_list.count("Maths"))
print()

print("******   extend()   *****   ")
print("Before operation :", var_list)
var_list.extend(["Leaf","Vitamins", "Maths"]) # similar to append but here you can append more tha onoe values
print("extend :", var_list)
print()

print("******   index()   *****   ")
print("index :", var_list.index("Leaf")) # Used to fetch the index
print("index :", var_list.index("Maths")) # returns first matched index value
print()

print("******   insert()   *****   ")
print("Before operation :", var_list)
var_list.insert(1, "Cat") # Used to insert the new values in the specific index place (before)
var_list.insert(1, ["Lion", "Pig"]) # can insert a new list also like nested list
print("insert :", var_list)
print()

print("******   pop()   *****   ")
print("Before operation :", var_list)
print("pop() :", var_list.pop()) # It removes an item and returns the same item (default is last index)
print("pop() :", var_list.pop())
print("pop(1) :", var_list.pop(1))
print("pop(5) :", var_list.pop(5))
print("After operation :", var_list)
print()

print("******   remove()   *****   ")
var_list.remove("h")
var_list.remove("Leaf") # Remove the mentoined value - first matched
print("After operation :", var_list)
print()

print("******   reverse()   *****   ")
var_list.reverse() # Used to reverse the order
print("After operation :", var_list)
print()

print("******   sort()   *****   ")
var_list_2 = [5,2,7,9,23,4,1,0,242]
print("Before operation :", var_list_2)
var_list_2.sort(reverse=False) # Used to sort the order (asc or desc)
print("After operation :", var_list_2)
var_list_2.sort(reverse=True)
print("After operation :", var_list_2)
print()
```

### 2. Set
1. One-Dimensional
2. Heterogenous
3. Wont allow duplicate values
4. Indexing and Slicing not possible
5. Mutable
6. empty vat = set()
7. Value assigning = {"abc", "xyz"}

```python
var_set = set()
#type(var_set)

# Assigning values
var_set = {"Chemistry","Maths","Physics","History","Drawing", "Maths"}
print("set() : ", var_set)
print()
###########################################################################
print("******   add()   *****   ")
var_set.add("Tamil") # Used to add a new value
#var_set.add(["qwe", "asd"]) # Multiple value or other list cannot be added
print("After operation :", var_set)
print()

print("******   clear()   *****   ")
var_set.clear() # wipe the values entirely
print("After operation :", var_set)
var_set = {"Chemistry","Maths","Physics","History","Drawing", "Biology"}
print(var_set)
print()

print("******   copy()   *****   ")
var_set_2 = var_set.copy() # Used to add a new value
print("copy :", var_set_2)
print()

print("******   difference()   *****   ")
var_set_3 = {"Chemistry","Maths","Physics","History","Drawing", "Tamil"}
var_set_4 = var_set.difference(var_set_3) # ued to get the different values which means not common in send set, it should return only left side diff
print("difference :", var_set_4)
print()

print("******   difference_update()   *****   ")
var_set.difference_update(var_set_3) # used to remove the common values from 2nd set into 1st set
print("difference_update :", var_set)
print()

print("******   discard()   *****   ")
print("Before operation :", var_set_2)
var_set_2.discard("Maths") # used to remove the value
print("After operation :", var_set_2)
print()

print("******   intersection()   *****   ")
print("Before operation set1:", var_set_2)
print("Before operation set2:", var_set_3)
var_set_6 = var_set_2.intersection(var_set_3) # union the common values
print("After operation set 3:", var_set_6)
print()

print("******   intersection_update()   *****   ")
print("Before operation set 1 :", var_set_2)
print("Before operation set 2:", var_set_3)
var_set_2.intersection_update(var_set_3) # update the set 1 f common values bet 2 sets
print("After operation set 1:", var_set_2)
print()

print("******   isdisjoint()   *****   ")
print("Before operation set 1:", var_set)
print("Before operation set 2:", var_set_2)
print("Before operation set 3:", var_set_3)
print("Before operation set 4:", var_set_4)
print("Before operation set 6:", var_set_6)
print("After operation set 1 & 2 :", var_set.isdisjoint(var_set_2)) # if 2 set doesn't have any common then True else False
print("After operation set 2 & 3:", var_set_2.isdisjoint(var_set_3))
print()

print("******   issubset()   *****   ")
print("Before operation set 1:", var_set_2)
print("Before operation set 2:", var_set_3)
print("issubset :", var_set_2.issubset(var_set_3)) # Return True if set A elements completely available in set B
var_set_2.add("English") 
print("Before operation set 1:", var_set_2)
print("Before operation set 2:", var_set_3)
print("issubset :", var_set_2.issubset(var_set_3))
var_set_2.discard("English") 
print()

print("******   issuperset()   *****   ")
print("Before operation set 1:", var_set_2)
print("Before operation set 2:", var_set_3)
print("issuperset :", var_set_2.issuperset(var_set_3)) # this is opp to issubset - Return True if set B(right) elements completely available in set A
var_set_2.add("English") 
print("Before operation set 1:", var_set_2)
print("Before operation set 2:", var_set_3)
print("issuperset :", var_set_2.issuperset(var_set_3))
print()

print("******   remove()   *****   ")
print("Before operation :", var_set_2)
print("remove():", var_set_2.remove("English")) # remove an item from set
#-- value must be be available otherwise thorws error
print("After operation set 1:", var_set_2)
print()

print("******   symmetric_difference()   *****   ")
var_set_2.add("English")
print("Before operation set 1:", var_set_2)
print("Before operation set 2:", var_set_3)
print("symmetric_difference :", var_set_2.symmetric_difference(var_set_3)) # Return the non common items from both set
print()

print("******   symmetric_difference_update()   *****   ")
var_set_2.add("English")
print("Before operation set 1:", var_set_2)
print("Before operation set 2:", var_set_3)
var_set_2.symmetric_difference_update(var_set_3) # Update set A with non common items from both set
print("After operation set 1:", var_set_2)
print()

print("******   union()   *****   ")
print("Before operation set 1:", var_set_2)
print("Before operation set 2:", var_set_3)
print("union():", var_set_2.union(var_set_3)) # retuns union all between 2 sets
#print("After operation :", var_set_2)
print()

print("******   update()   *****   ")
print("Before operation set 1:", var_set_2)
print("Before operation set 2:", var_set_3)
var_set_2.update(var_set_3) # updates set A with union all between 2 sets
print("After operation set 1:", var_set_2)
print()
```

### 3. Tuple
- One Dimensional
- Heterogenous
- Allow Duplicate
- Allow slicing
- Immutable
- Empty var = ()
- Value assiging = ("abc", "xyz")

```python
var_tuple_1 = ()
#type(var_tuple_1)
# Assigning values
var_tuple_1 = ("Apple","Mango","Apple", 10, 20, -5, 10)
print("var_tuple_1 :", var_tuple_1)
print()

#########################################################
print("******   count()   *****   ")
print("count (Mango):", var_tuple_1.count("Mango")) # Used to count the no. of occurence of Value
print("count (Apple):", var_tuple_1.count("Apple"))
print("count (10):", var_tuple_1.count(10))
print()

print("******   index()   *****   ")
# Used to return index value of value and we can set starting index (default first matched index will retun)
print("index (Mango):", var_tuple_1.index("Mango")) 
print("index (Apple):", var_tuple_1.index("Apple"))
print("index (Apple, 1):", var_tuple_1.index("Apple", 1))
print("index (10):", var_tuple_1.index(10))
print("index (10, 4):", var_tuple_1.index(10,4))
print()
```

### 4. Dictionary
- Two Dimensional
- Heterogenous (accepts Int and string)
- Indexing and Slicing not possible
- Mutable
- Duplicate key wont allows but values allows
- Empty var = {}
- Value assiging = {"abc" : 54, "xyz": 13}

```python
var_dict_1 = {}
#type(var_dict_1)
# Assigning values
var_dict_1 = {"Maths": 98, "Chemistry": 86, "Physics":56.6, "Maths": 98.3}
print("var_dict_1",var_dict_1)
print()
##########################################################################

print("******   clear()   *****   ")
print("Before operation dict 1:", var_dict_1)
var_dict_1.clear() # wipe the elemtns entirely
print("After operation dict 1:", var_dict_1)
print()
var_dict_1 = {"Maths": 98, "Chemistry": 86, "Physics":56.6, "Maths": 98.3, "Tamil": 99.3, "Drawing": 50}

print("******   copy()   *****   ")
#print("Before operation dict 1:", var_dict_1)
print("copy :", var_dict_1.copy()) # return a clone of set
##print("After operation dict 1:", var_dict_1)
print()

print("******   fromkeys()   *****   ")
print("fromkeys default:", dict.fromkeys(var_dict_1)) # return only keys and defalt value is none
print("fromkeys custom value:", dict.fromkeys(var_dict_1, 10)) # return only keys and defalt value is none
print()

print("******   get()   *****   ")
print("get :", var_dict_1.get("Maths")) # return a value of a specified key if key not avail then None
print("get :", var_dict_1.get("Maths_1")) 
print()

print("******   items()   *****   ")
print("items :", var_dict_1.items()) # return a key value pairs - seems useful for loping
print()

print("******   keys()   *****   ")
print("keys :", var_dict_1.keys()) # return a only keys
print()

print("******   pop()   *****   ")
print("Before operation dict 1:", var_dict_1)
print("pop :", var_dict_1.pop("Drawing")) # it removes the specifed key and returns value of the key - update also happened to var - key must be avail
print("After operation dict 1:", var_dict_1)
print()

print("******   popitem()   *****   ")
print("Before operation dict 1:", var_dict_1)
print("popitem :", var_dict_1.popitem()) # it removes the last key and returns the key & value - update also happened to var
print("After operation dict 1:", var_dict_1)
print()

print("******   setdefault()   *****   ")
print("Before operation dict 1:", var_dict_1)
print("setdefault :", var_dict_1.setdefault("Maths_1", 100)) # if key matches with dict then return the value if key not avial the inserts
print("After operation dict 1:", var_dict_1)
print()

print("******   update()   *****   ")
print("Before operation dict 1:", var_dict_1)
var_dict_1.update({"Maths_2": 100}) 
# directly updates dict if key matches if not insert - here we can provide multiple 
var_dict_1.update({"Maths": 100})
print("After operation dict 1:", var_dict_1)
print()

print("******   values()   *****   ")
print("values :", var_dict_1.values()) # retun only values
print()
```
