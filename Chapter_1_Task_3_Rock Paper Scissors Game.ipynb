# Task 3 - Rock Paper Scissors Game

```python
import random

print("Welcome to Rock, Paper, Scissors Game !!!")

def get_user_input():
    while True:
        try:
            user_input = int(input("Enter your option: 1-Rock, 2-Paper, 3-Scissors : "))

            if user_input >= 1 and user_input <= 3:
                return user_input
            else:
                print("User input is Invalid, please choose between 1-3 (1-Rock, 2-Paper, 3-Scissors)")
        except ValueError:
            print("Input Not recognized, please choose between 1-3 (1-Rock, 2-Paper, 3-Scissors)")

def play_game():
    while True:
        options = {1:"Rock", 2:"Paper", 3:"Scissors"} 
        
        system_selection = random.randint(1,3)
        user_selection = get_user_input()
    
        if user_selection == system_selection:
            print("User selection is {} and System selection is {}".format(options.get(user_selection), options.get(system_selection)))
            print("Tie")
            print()
        elif (user_selection == 1 and system_selection == 3) or (user_selection == 2 and system_selection == 1) or (user_selection == 3 and system_selection == 2):
            print("User selection is {} and System selection is {}".format(options.get(user_selection), options.get(system_selection)))
            print("You win !!!!!!!!!!!")
            print()
            return "u"
        else:
            print("User selection is {} and System selection is {}".format(options.get(user_selection), options.get(system_selection)))
            print("System win !!!")
            print()
            return "s"
        
        #print("system_choice: ", options.get(system_selection))
        #print("user_choice: ", options.get(user_selection))
        #print("User selection is {} and System selection is {}".format())


def get_exit_input():
    while True:
        try:
            user_exit_input = int(input("Enter your option: 1-Play Again, 2-Exit :: "))

            if user_exit_input >= 1 and user_exit_input <= 2:
                return user_exit_input
            else:
                print("User input is Invalid, please choose between 1-2 (1-Play Again, 2-Exit)")
        except ValueError:
            print("Input Not recognized, please choose between 1-2 (1-Play Again, 2-Exit)")
        
max_point = 5
user_iteration = system_iteration = 0

while True:
    won_by = play_game()

    if (won_by == "u"):
        user_iteration += 1
    else:
        system_iteration += 1

    print("Your Points: {}\nSystem Points: {}".format(user_iteration, system_iteration))
          
    if user_iteration == 5:
        print("Game Over - You won !!!")
        user_exit_input = get_exit_input()
        user_iteration = system_iteration = 0
        if user_exit_input == 2:
            print("Have a nice day !!!")
            break
    elif system_iteration == 5:
        print("Game Over - System won !!!")
        user_iteration = system_iteration = 0
        user_exit_input = get_exit_input()
        if user_exit_input == 2:
            print("Have a nice day !!!")
            break
    print()
    
    '''if user_exit_input == 1:
        continue
    else:
        print("Have a nice day !!!")
        break  
'''
        
```
