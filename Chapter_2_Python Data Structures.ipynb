# Why Python Built-In Data Structures (DS)

* To collect multiple data in a structured way.

### 4 Types of DS in Python
1. List
2. Sets
3. Tuples
4. Dictionary - Key & Value

## 1. List
- 1D
- Heterogenous
- Accepts Duplicates values
- Sliced
- Mutable
- mentioned in []

```python
empty_cart = []
```

```python
empty_cart
```

```python
type(empty_cart)
```

```python
items_to_buy = ["Watermelon","Kiwi","Peach", "Egg", "Kiwi", 6.98, 45]
items_to_buy

#Assigning List and printing
```

```python
type(items_to_buy)
# Gather the type
```

```python
len(items_to_buy)
#Checking length
```

```python
items_to_buy
```

```python
items_to_buy[2]
```

```python
items_to_buy[-3]
```

```python
items_to_buy[1:5]
#Slicing is nothing but we can fetch n to n values from List
```

```python
items_to_buy[-3:-1]
#Reverse values consideration also possible (Index starts with -1 but for noraml case starts with 0)
```

```python
items_to_buy[5:2]
#Invalid order for regular case
```

```python
items_to_buy[2] = "Mango"
#Trying mutation
```

```python
items_to_buy
```

## 2. Set

1. One-Dimensional
2. Heterogenous
3. Wont allow duplicate values
4. Indexing and Slicing not possible
5. Mutable
6. mentioned in set()

```python
empty_cart_2 = set()
#Creating empty set instance
```

```python
empty_cart_2
```

```python
type(empty_cart_2)
```

```python
items_to_buy_2 = {"Watermelon","Kiwi","Peach", "Egg", "Kiwi", 45, 6.98, 45}
items_to_buy_2
#Assigning Values inside {}
```

```python
items_to_buy_2[2]
#Indexing not possible
```

```python
items_to_buy_2[2] = "Milk"
#Mutation is not by indexing
```

```python
items_to_buy_2.add("Banana")
#Adding a new value so mutable
#items_to_buy_2.update("Egg1", "Chicken")
print(items_to_buy_2)
```

## 3. Tuples
- One Dimensional
- Heterogenous
- Allow Duplicate
- Allow slicing
- Immutable
- mentioned in ()

```python
empty_cart_3 = ()
empty_cart_3
```

```python
items_to_buy_3 = ("Watermelon","Kiwi","Peach", "Egg", "Kiwi", 45, 6.98, 45)
items_to_buy_3
#Assiging and printing
```

```python
type(items_to_buy_3)
```

```python
# Indexing
items_to_buy_3[5]
```

```python
# Slicing
items_to_buy_3[2:5]
```

```python
# Checking Mutable
items_to_buy_3[2] = "Potato"
#So not mutable
```

```python
#Count
items_to_buy_3.count("Kiwi")
```

```python
# Count
items_to_buy_3.count(items_to_buy_3[1])
```

## 4. Dictionary
It's a Key:Value pair

- Two Dimensional
- Heterogenous (accepts Int and string)
- Indexing and Slicing not possible
- Mutable
- Duplicate key wont allows but values allows
- mentioned in {}

```python
# Creating empty instance
empty_cart_4 = {}
empty_cart_4
```

```python
#Checking type
type(empty_cart_4)
```

```python
# Stores value
students_mark = {"Maths": 98, "Chemistry": 86, "Physics":56.6, "Maths": 98.3}
students_mark
```

```python
type(students_mark)
```

```python
# Indexing

students_mark[1]
```

```python
# Slicing 
students_mark[0:1]
```

```python
# Generate only keys
students_mark.keys()
```

```python
# Generate Values
students_mark.values()
```

```python
import pandas as pd
```

```python
pd.DataFrame(data = students_mark, index=0)
```

```python
students_mark_new = {"Names":["Ben", "Leon","Sai", "Catrin"], 
                     "Maths":[68,63,89,57],
                     "Science": [99, 55, 66, 77]}
students_mark_new
```

```python
pd.DataFrame(data = students_mark_new)
```
