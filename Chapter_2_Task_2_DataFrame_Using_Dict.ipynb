```python
import pandas as pd
var_dict_1 = {}
#type(var_dict_1)
# Assinging values

var_dict_1 = {"Names":["Ben", "Leon","Sai", "Catrin"], 
              "Age":[45,12,36,30],
              "Gender": ["M", "M", "M", "F"],
              "GPD": [8.9, 7.2, 5.5, 9.8]}

pd.DataFrame(var_dict_1)
```

```python
var_dict_2 = {
                "Name" : {0: "Arjun", 1: "Priya", 2: "Ravik"},
                "Age"  : {0: 38, 1: 25, 2: 32},
                "City" : {0: "Berlin", 1: "Schwetzingen", 2: "Munich"}
             }
#type(var_dict_2)

pd.DataFrame(var_dict_2)
```
